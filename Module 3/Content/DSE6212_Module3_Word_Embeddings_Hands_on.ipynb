```python
# Load Pre-Trained Word2Vec Embeddings with Gensim

from gensim.models import KeyedVectors
from gensim.models import KeyedVectors
import gensim.downloader as api

# Load pre-trained Word2Vec model from Gensim
word2vec_model = api.load("word2vec-google-news-300")

# Explore the vocabulary
print("Vocabulary Size:", len(word2vec_model.key_to_index))

# Example: Get vector for a word
word_vector = word2vec_model['king']
print("\nVector for 'king':", word_vector[:10])  # Print the first 10 dimensions

# Example: Most similar words to 'king'
print("\nMost Similar Words to 'king':")
print(word2vec_model.most_similar('king'))
```

Loading Pre-trained GloVe Embeddings

```python
!wget https://nlp.stanford.edu/data/glove.6B.zip # Download the GloVe embeddings
!unzip glove.6B.zip # Unzip the downloaded archive

import numpy as np

# Load GloVe embeddings
# glove_path = "https://nlp.stanford.edu/data/glove.6B.zip" # This line is no longer needed
# Note: Download and extract the embeddings first # This comment is no longer needed

# Load the 100-dimensional embeddings
embeddings_index = {}
with open("glove.6B.100d.txt", encoding="utf-8") as f:
    for line in f:
        values = line.split()
        word = values[0]
        coefs = np.asarray(values[1:], dtype='float32')
        embeddings_index[word] = coefs

print("Vocabulary Size:", len(embeddings_index))

# Example: Get vector for a word
print("\nVector for 'queen':", embeddings_index['queen'])
```

Generalizing Embeddings with Keras

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Example corpus
texts = ["The quick brown fox jumps over the lazy dog",
         "I love natural language processing with word embeddings"]

# Tokenize and pad sequences
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)
word_index = tokenizer.word_index
sequences = tokenizer.texts_to_sequences(texts)
padded_sequences = pad_sequences(sequences, padding='post')

print("Word Index:", word_index)
print("\nPadded Sequences:\n", padded_sequences)

# Create an Embedding Layer
vocab_size = len(word_index) + 1
embedding_dim = 50
embedding_layer = Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=padded_sequences.shape[1])

# Get embeddings for the input sequences
embedded_sequences = embedding_layer(padded_sequences)
print("\nEmbedded Sequences Shape:", embedded_sequences.shape)
```

### Cosine Similarity for Word Relationships

```python
from sklearn.metrics.pairwise import cosine_similarity

# Example: Cosine similarity between two words using Word2Vec
vector_king = word2vec_model['king']
vector_queen = word2vec_model['queen']
similarity = cosine_similarity([vector_king], [vector_queen])
print("\nCosine Similarity between 'king' and 'queen':", similarity[0][0])

# Example: Cosine similarity using GloVe
vector_water = embeddings_index['water']
vector_steam = embeddings_index['steam']
similarity = cosine_similarity([vector_water], [vector_steam])
print("\nCosine Similarity between 'water' and 'steam':", similarity[0][0])
```
